# Production Model Analysis

Audit the currently deployed model - metrics, decision boundary, feature importance, and version history.
Use this to understand what you're competing against before building a challenger.

**What this notebook covers:**
- Load production model from MLflow registry
- Evaluate all 6 metrics on the latest batch inference ground truth
- Confusion matrix, ROC curve, Precision-Recall curve, calibration
- SHAP feature importance
- Model version history

````python
import os
import sys
import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import mlflow
from sqlalchemy import create_engine

sys.path.insert(0, '/opt/mlops/services')
from metrics import REGISTRY, compute_all_metrics, display_name

POSTGRES_HOST = os.getenv('POSTGRES_HOST', 'postgres')
POSTGRES_PORT = os.getenv('POSTGRES_PORT', '5432')
POSTGRES_DB   = os.getenv('POSTGRES_DB', 'mlops')
POSTGRES_USER = os.getenv('POSTGRES_USER', 'mlops_user')
POSTGRES_PASS = os.getenv('POSTGRES_PASSWORD', '')
MLFLOW_URI    = os.getenv('MLFLOW_TRACKING_URI', 'http://mlflow:5000')
MODEL_NAME    = os.getenv('MODEL_REGISTRY_NAME', 'credit-risk-classifier')

engine = create_engine(
    f'postgresql+psycopg2://{POSTGRES_USER}:{POSTGRES_PASS}'
    f'@{POSTGRES_HOST}:{POSTGRES_PORT}/{POSTGRES_DB}'
)
mlflow.set_tracking_uri(MLFLOW_URI)

FEATURE_COLUMNS = [
    'age', 'annual_income', 'credit_score', 'loan_amount',
    'loan_term_months', 'employment_length_years', 'home_ownership_encoded',
    'debt_to_income_ratio', 'num_credit_lines', 'payment_history_score',
]

print('Ready.')
````

## 1  Load Production Model

````python
client = mlflow.MlflowClient()
prod_version = client.get_model_version_by_alias(MODEL_NAME, 'production')

primary_metric = prod_version.tags.get('primary_metric', 'roc_auc')
print(f'Model:          {MODEL_NAME}')
print(f'Version:        {prod_version.version}')
print(f'Run ID:         {prod_version.run_id}')
print(f'Primary metric: {display_name(primary_metric)} ({primary_metric})')

# Load model
prod_model = mlflow.pyfunc.load_model(f'models:/{MODEL_NAME}@production')
print(f'\nModel loaded: {type(prod_model._model_impl).__name__}')
````

## 2  Load Evaluation Data

````python
# Use the most recent batch run that has labelled ground truth
eval_df = pd.read_sql("""
    SELECT p.run_index, p.record_id, p.customer_id,
           p.default_probability, p.default_flag_predicted,
           COALESCE(p.actual_default_flag, c.default_flag) AS actual_default_flag,
           p.age, p.annual_income, p.credit_score, p.loan_amount,
           p.loan_term_months, p.employment_length_years, p.home_ownership_encoded,
           p.debt_to_income_ratio, p.num_credit_lines, p.payment_history_score
    FROM dwh_history.prediction_ground_truth p
    LEFT JOIN dwh_clean.cleaned_features c ON p.record_id = c.record_id
    WHERE COALESCE(p.actual_default_flag, c.default_flag) IS NOT NULL
    ORDER BY p.run_index DESC, p.predicted_at DESC
    LIMIT 5000
""", engine)

if eval_df.empty:
    print('No labelled ground truth available yet.')
    print('Run: Control Panel → Batch Inference → then re-run this cell.')
else:
    run_indices = sorted(eval_df['run_index'].unique(), reverse=True)
    print(f'Labelled records:  {len(eval_df):,}')
    print(f'Run indices:       {run_indices}')
    print(f'Default rate:      {eval_df["actual_default_flag"].mean():.3f}')
````

## 3  All Metrics

````python
y_true  = eval_df['actual_default_flag'].astype(int).values
y_score = eval_df['default_probability'].astype(float).values
y_pred  = (y_score >= 0.5).astype(int)

metrics = compute_all_metrics(y_true, y_score)

print(f'Evaluation on {len(y_true):,} labelled records\n')
for key, label in REGISTRY.items():
    star = ' ◀ primary' if key == primary_metric else ''
    print(f'  {label:<12} {metrics[key]:.4f}{star}')
````

## 4  ROC, Precision-Recall & Calibration Curves

````python
from sklearn.metrics import roc_curve, precision_recall_curve, roc_auc_score, average_precision_score

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# ROC
fpr, tpr, _ = roc_curve(y_true, y_score)
auc = roc_auc_score(y_true, y_score)
axes[0].plot(fpr, tpr, color='#ef4444', lw=2.5, label=f'AUC = {auc:.4f}')
axes[0].fill_between(fpr, tpr, alpha=0.08, color='#ef4444')
axes[0].plot([0, 1], [0, 1], 'k--', lw=1.5, label='No skill')
axes[0].set_xlabel('False Positive Rate')
axes[0].set_ylabel('True Positive Rate')
axes[0].set_title('ROC Curve', fontweight='bold')
axes[0].legend()

# Precision-Recall
prec, rec, _ = precision_recall_curve(y_true, y_score)
ap = average_precision_score(y_true, y_score)
baseline = y_true.mean()
axes[1].plot(rec, prec, color='#22c55e', lw=2.5, label=f'AP = {ap:.4f}')
axes[1].fill_between(rec, prec, alpha=0.08, color='#22c55e')
axes[1].axhline(baseline, color='#94a3b8', linestyle='--', lw=1.5, label=f'Baseline ({baseline:.3f})')
axes[1].set_xlabel('Recall')
axes[1].set_ylabel('Precision')
axes[1].set_title('Precision-Recall Curve', fontweight='bold')
axes[1].legend()

# Calibration
try:
    from sklearn.calibration import calibration_curve
    frac_pos, mean_pred = calibration_curve(y_true, y_score, n_bins=10)
    axes[2].plot(mean_pred, frac_pos, 's-', color='#8b5cf6', lw=2, ms=7, label='Model')
except Exception:
    axes[2].text(0.5, 0.5, 'calibration_curve unavailable', ha='center', va='center')
axes[2].plot([0, 1], [0, 1], 'k--', lw=1.5, label='Perfectly calibrated')
axes[2].set_xlabel('Mean Predicted Probability')
axes[2].set_ylabel('Fraction of Positives')
axes[2].set_title('Calibration', fontweight='bold')
axes[2].legend()

plt.tight_layout()
plt.show()
````

## 5  Confusion Matrix & Threshold Sensitivity

````python
from sklearn.metrics import confusion_matrix

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Confusion matrix at 0.5
cm = confusion_matrix(y_true, y_pred)
total = cm.sum()
im = axes[0].imshow(cm, cmap='Blues')
for i in range(2):
    for j in range(2):
        axes[0].text(j, i, f'{cm[i,j]}\n({cm[i,j]/total:.1%})',
                     ha='center', va='center', fontsize=13,
                     color='white' if cm[i,j] > cm.max()*0.6 else 'black')
axes[0].set_xticks([0, 1]); axes[0].set_xticklabels(['Pred: No Default', 'Pred: Default'])
axes[0].set_yticks([0, 1]); axes[0].set_yticklabels(['Actual: No Default', 'Actual: Default'])
axes[0].set_title('Confusion Matrix  (threshold = 0.5)', fontweight='bold')

# Threshold sensitivity
thresholds = np.linspace(0.1, 0.9, 80)
f1s, precs, recs = [], [], []
from sklearn.metrics import f1_score, precision_score, recall_score
for t in thresholds:
    yp = (y_score >= t).astype(int)
    f1s.append(f1_score(y_true, yp, zero_division=0))
    precs.append(precision_score(y_true, yp, zero_division=0))
    recs.append(recall_score(y_true, yp, zero_division=0))

axes[1].plot(thresholds, f1s,   label='F1',        color='#3b82f6', lw=2)
axes[1].plot(thresholds, precs, label='Precision',  color='#f59e0b', lw=2)
axes[1].plot(thresholds, recs,  label='Recall',     color='#22c55e', lw=2)
axes[1].axvline(0.5, color='#94a3b8', linestyle='--', lw=1.5, label='Current threshold')
best_t = thresholds[np.argmax(f1s)]
axes[1].axvline(best_t, color='#3b82f6', linestyle=':', lw=1.5, label=f'Best F1 threshold ({best_t:.2f})')
axes[1].set_xlabel('Decision Threshold')
axes[1].set_ylabel('Score')
axes[1].set_title('Threshold Sensitivity', fontweight='bold')
axes[1].legend(fontsize=9)
axes[1].set_ylim(0, 1)

plt.tight_layout()
plt.show()
print(f'Best F1 threshold: {best_t:.2f}  →  F1={max(f1s):.4f}  (current threshold 0.5 → F1={f1s[list(thresholds).index(min(thresholds, key=lambda t: abs(t-0.5)))]:.4f})')
````

## 6  SHAP Feature Importance

````python
import shap

X_eval = eval_df[FEATURE_COLUMNS].head(500)  # cap at 500 for speed

def _get_underlying(model):
    impl = model._model_impl
    for attr in ('sklearn_model', 'xgb_model', '_model'):
        if hasattr(impl, attr):
            return getattr(impl, attr)
    return impl

underlying = _get_underlying(prod_model)
print(f'Underlying model type: {type(underlying).__name__}')

try:
    explainer = shap.TreeExplainer(underlying)
    shap_out = explainer(X_eval)
    sv = shap_out.values
    if sv.ndim == 3:
        sv = sv[:, :, 1]
    explainer_type = 'TreeExplainer'
except Exception as e:
    print(f'TreeExplainer failed ({e}), using LinearExplainer/KernelExplainer fallback')
    background = shap.sample(X_eval, min(50, len(X_eval)))
    def _predict(arr):
        df = pd.DataFrame(arr, columns=FEATURE_COLUMNS)
        out = prod_model.predict(df)
        return out.values if hasattr(out, 'values') else np.array(out)
    explainer = shap.KernelExplainer(_predict, background)
    sv = explainer.shap_values(X_eval.values[:50], nsamples=50)
    if isinstance(sv, list):
        sv = sv[1]
    explainer_type = 'KernelExplainer'

mean_abs = {feat: float(np.abs(sv[:, i]).mean()) for i, feat in enumerate(FEATURE_COLUMNS)}
fi_df = pd.Series(mean_abs).sort_values(ascending=True)

fig, ax = plt.subplots(figsize=(9, 5))
colors = ['#3b82f6' if v > fi_df.median() else '#93c5fd' for v in fi_df.values]
ax.barh(fi_df.index, fi_df.values, color=colors)
ax.set_xlabel('Mean |SHAP value|')
ax.set_title(f'Feature Importance ({explainer_type})', fontweight='bold')
for i, (feat, val) in enumerate(fi_df.items()):
    ax.text(val + 0.0002, i, f'{val:.4f}', va='center', fontsize=9)
plt.tight_layout()
plt.show()

print(f'\nTop features:')
for feat, val in sorted(mean_abs.items(), key=lambda x: -x[1]):
    print(f'  {feat:<32} {val:.4f}')
````

## 7  Model Version History

````python
client = mlflow.MlflowClient()
versions = client.search_model_versions(f"name='{MODEL_NAME}'")
versions = sorted(versions, key=lambda v: int(v.version), reverse=True)

rows = []
for v in versions:
    run = None
    try:
        run = client.get_run(v.run_id)
    except Exception:
        pass
    rows.append({
        'version':        v.version,
        'stage':          v.tags.get('stage', '-'),
        'primary_metric': v.tags.get('primary_metric', '-'),
        'run_id':         v.run_id[:20] + '…',
        'roc_auc':        round(run.data.metrics.get('test_roc_auc', 0), 4) if run else '-',
        'f1':             round(run.data.metrics.get('test_f1', run.data.metrics.get('test_f1_score', 0)), 4) if run else '-',
        'created':        pd.Timestamp(v.creation_timestamp, unit='ms').strftime('%Y-%m-%d %H:%M'),
    })

hist_df = pd.DataFrame(rows)
print(f'Registry: {len(versions)} version(s) of {MODEL_NAME}\n')
hist_df
````

## 8  Metric Trend Across Monitoring Runs

````python
trend = pd.read_sql("""
    SELECT run_index, roc_auc, pr_auc, f1_score, precision_score,
           recall_score, accuracy, primary_metric, retraining_triggered
    FROM dwh_monitoring_hard.results
    ORDER BY run_index
""", engine)

if trend.empty:
    print('No monitoring runs yet. Run Hard Metrics monitoring from the Control Panel.')
else:
    fig, ax = plt.subplots(figsize=(12, 5))
    series = [
        ('roc_auc',         'ROC-AUC',   '#ef4444'),
        ('pr_auc',          'PR-AUC',    '#f97316'),
        ('f1_score',        'F1',        '#22c55e'),
        ('precision_score', 'Precision', '#8b5cf6'),
        ('recall_score',    'Recall',    '#f59e0b'),
        ('accuracy',        'Accuracy',  '#3b82f6'),
    ]
    for col, label, color in series:
        if col in trend.columns:
            ax.plot(trend['run_index'], trend[col], 'o-', label=label, color=color, lw=2, ms=6)

    retrain_runs = trend[trend.get('retraining_triggered', False) == True]['run_index'].tolist() \
        if 'retraining_triggered' in trend.columns else []
    for ri in retrain_runs:
        ax.axvline(ri, color='#94a3b8', linestyle=':', lw=1.5)
        ax.text(ri + 0.05, 0.1, 'retrain', fontsize=8, color='#64748b', rotation=90)

    ax.set_xlabel('Batch Run Index')
    ax.set_ylabel('Score')
    ax.set_ylim(0, 1.05)
    ax.set_title('All Metrics Across Monitoring Runs', fontweight='bold')
    ax.legend(ncol=3, fontsize=9)
    ax.xaxis.set_major_locator(mticker.MaxNLocator(integer=True))
    plt.tight_layout()
    plt.show()

    print(f'\nLatest monitoring run ({int(trend["run_index"].max())}):')
    latest = trend.iloc[-1]
    for col, label, _ in series:
        val = latest.get(col)
        if val is not None:
            star = ' ◀ primary' if col.replace('_score','').replace('_',  '') == str(latest.get('primary_metric','')).replace('_','') or col == latest.get('primary_metric') else ''
            print(f'  {label:<12} {float(val):.4f}{star}')
````
